# MIMII Anomalib DataModule

This notebook uses a custom Lightning DataModule that reads `data/dcase-2020-spectrogram/meta.json` generated by  `mimii-toy.py` script.

```python
import importlib
import sys
from pathlib import Path
from timm.data.constants import IMAGENET_DEFAULT_MEAN
from timm.data.constants import IMAGENET_DEFAULT_STD
from anomalib.engine import Engine
from torchvision.transforms.v2 import Compose, Resize, ToTensor, Normalize, CenterCrop

repo_root = Path.cwd()
if not (repo_root / "src").exists():
    repo_root = repo_root.parent
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))

import src.mimii_anomalib_datamodule as mimii_dm_module
importlib.reload(mimii_dm_module)
MIMIIAnomalibDataModule = mimii_dm_module.MIMIIAnomalibDataModule
```

```python
dm = MIMIIAnomalibDataModule(
    root="../data/dcase-2020-three-channel",
    categories=("fan", "pump", "slider", "valve"),
    train_phases=("id_00",),
    val_phases=("id_04",),
    test_phases=("id_02",),
    train_batch_size=32,
    eval_batch_size=32,
    num_workers=8,
    augmentations=Compose([
        Resize((256, 256)),
        ToTensor(),
        Normalize(mean=IMAGENET_DEFAULT_MEAN, std=IMAGENET_DEFAULT_STD, inplace=True),
		]
    ),
)

dm.setup("fit")
train_batch = next(iter(dm.train_dataloader()))
print("train keys:", train_batch.keys(include_none=False))
print("train image shape:", train_batch.image.shape)
print("train label unique:", train_batch.gt_label.unique())
```

```python
from src.mvtecad_patched_datamodule import MVTecADPatched
dm = MVTecADPatched(
	root="../data/mvtec",
	train_batch_size=32,
	eval_batch_size=32,
	num_workers=8,
	categories=(
		"bottle",
		"cable",
		"capsule",
		"carpet",
		"grid",
		"hazelnut",
		"leather",
		"metal_nut",
		"pill",
		"screw",
		"tile",
		"toothbrush",
		"transistor",
		"wood",
		"zipper",
	),
	augmentations=Compose([
		Resize((256, 256)),
		CenterCrop((256, 256)),
		ToTensor(),
		Normalize(mean=IMAGENET_DEFAULT_MEAN, std=IMAGENET_DEFAULT_STD, inplace=True),
	]),
)
```

```python
from anomalib.models import Padim
model = Padim(backbone="resnet18", n_features=100) # according to the paper
engine = Engine()
engine.fit(model=model, datamodule=dm)
engine.test(model=model, datamodule=dm)
```

```python
from anomalib.models import Padim
import os
os.environ["CUDA_VISIBLE_DEVICES"] = "2"
model = Padim(backbone="wide_resnet50_2", n_features=550) # according to the paper
engine = Engine()
engine.fit(model=model, datamodule=dm)
engine.test(model=model, datamodule=dm)
```

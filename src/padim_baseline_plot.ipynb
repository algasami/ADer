```python
import matplotlib.pyplot as plt
import numpy as np
import csv

def plot_by_metric(metric: str):

	with open("padim_performance.csv", "r") as f:
		reader = csv.DictReader(f)
		models = []
		data = []
		for row in reader:
			model_name = row["model"]
			models.append(model_name)
			data.append(float(row[metric]))
		
		models, data = zip(*sorted(zip(models, data), key=lambda x: x[1], reverse=False))
		plt.barh(models, data)
		plt.xlabel(metric)
		plt.title(f"{metric}")
		plt.xlim(0, 1)
		plt.grid(axis="x", linestyle="--", alpha=0.7)
		plt.tight_layout()
		plt.show()

# plot_by_metric(["image_AUROC", "image_F1Score", "pixel_AUROC", "pixel_F1Score"])
for metric in ["image_AUROC", "image_F1Score", "pixel_AUROC", "pixel_F1Score"]:
	plot_by_metric(metric)
```

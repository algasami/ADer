````python
import glob, os
import librosa 
import numpy as np
import cv2
from scipy.io import wavfile
from matplotlib import pyplot as plt
import wave
````

````python
# define your own dataset root

root = './dataset/'
````

````python
cats = ['fan','valve','pump','slider']
subs = ['id_00', 'id_02', 'id_04', 'id_06']
labels = ['normal', 'abnormal']
````

````python
def get_paths(cat, sub, label, suffix='out'):
    path = root + cat + '/' + sub + '/' + label + '/'
    targetpath = root + cat + '/' + sub + '_' + suffix + '/' + label + '/'
    
    print(path)
    print(targetpath)
    
    if not os.path.exists(path):
        raise FileNotFoundError("path not found")
    if not os.path.exists(targetpath):
        print("new target path created.")
        os.makedirs(targetpath)

    return path, targetpath
````

````python
def scale_minmax(X, smin=0.0, smax=1.0):
    X_std = (X - X.min()) / (X.max() - X.min())
    X_scaled = X_std * (smax - smin) + smin
    return X_scaled

def save_mel_img(fname, targetpath, hop_length=512, n_mels=128, fmax=8000):
    y, sr = librosa.load(fname, sr=None)
    S = librosa.feature.melspectrogram(y=y, 
                                       sr=sr, 
                                       n_mels=n_mels, 
                                       n_fft=hop_length*2, 
                                       hop_length=hop_length, 
                                       fmax=fmax,
                                      )
    S = librosa.power_to_db(S, ref=np.max)
    img = scale_minmax(S, 0, 255).astype(np.uint8)
    img = np.flip(img, axis=0)
    
    # plt.figure(figsize=(4,2))
    # plt.imshow(img)
    # plt.show()
    
    basename = os.path.basename(fname)
    savename = targetpath + basename[:-4] + '.png'
    print(savename, S.shape, y.shape[0])
    cv2.imwrite(savename, img)
````

````python
%%time

for cat in cats:
    for sub in subs:
        for label in labels:
            path, targetpath = get_paths(cat, sub, label)
            fnames = glob.glob(path+'*.wav')
            print("folder size:", len(fnames))

            for fname in fnames:
                save_mel_img(fname, targetpath)
````
